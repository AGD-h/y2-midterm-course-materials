```python
import pandas as pd
import numpy as np
from plotnine import *
import matplotlib.pyplot as plt
plt.rc('font',family='Times New Roman')
N=20
df1 =pd.DataFrame(dict(x=np.sort(np.random.randn(N)),y=np.sort(np.random.randn(N))))
df2 =pd.DataFrame(dict(x=df1.x+0.3*np.sort(np.random.randn(N)),y=df1.y+0.1*np.sort(np.random.randn(N))))

#--------------1. All layers share the same data source and aesthetic mapping parameters.---------------------------
p1=(ggplot(df1,aes('x','y',colour='x+y'))+
       geom_line(size=1)+
       geom_point(shape='o',size=5)+
       scale_color_distiller(name='Line',palette='Blues')+
       guides(color=guide_colorbar(title='Point\nLine'))+
       theme(text=element_text(size=12,color='k'), aspect_ratio=1.25,dpi=100,figure_size=(4,4)))
p1.show()
#p1.save('plotnine_Advance1.pdf')
```

```python
#-----------------------------2. All layers share only the data source.--------------------------------------------
p2=(ggplot(df1,aes('x','y'))+
        geom_line(aes(color='x+y'),size=1)+
        geom_point(aes(fill='x+y'),color='k',shape='o',size=5)+
        scale_fill_distiller(name='Point',palette="YlOrRd")+
        scale_color_distiller(name='Line',palette='Blues')+
        theme(text=element_text(size=12,color='k'),
                   aspect_ratio=1.25,dpi=100,figure_size=(4,4)))
p2.show()
```

```python
#----------------3. Each layer object uses independent data sources and aesthetic mapping parameters.------------
p3=(ggplot()+
        geom_line(aes('x','y',color='x+y'),df1,size=1)+
        geom_point(aes('x','y',fill='x+y'),df2,color='k', shape='o',size=5)+
        scale_fill_distiller(name='Point',palette='YlOrRd')+
        scale_color_distiller(name='Line',palette='Blues')+
        theme(text=element_text(size=12,color='k'),
                   aspect_ratio=1.25,dpi=100,figure_size=(4,4)))
p3.show()
#p3.save('plotnine_Advance3.pdf')
```

```python
import pandas as pd
import numpy as np
from plotnine import *
df=pd.read_csv("MappingAnalysis_Data.csv")
#--------------------------(a)Position +Direction------------------------------
p1=(ggplot(df,aes(x='Time',y='value',group='variable'))+
       geom_line()+
       geom_point(shape='o',size=4,color='k',fill='white')+
       theme_classic()+
       theme(text=element_text(size=12,color='k'),aspect_ratio=0.8,
                  dpi=100,  figure_size=(4,4)))
p1.show()
#p1.save(plotnine(a).pdf)
```

```python
#------------------(b) Position + Direction+Saturation-----------------------------------
p2=(ggplot(df,aes(x='Time',y='value',fill='variable'))+
       geom_line()+
       geom_point(shape='o',size=4,color='k')+
       scale_fill_manual(values=('#595959','#BFBFBF','black','white'))+
       theme_classic()+
       theme(text=element_text(size=12,color='k'),aspect_ratio=0.8,
                  dpi=100, figure_size=(4,4),
                  legend_position=(0.12,0.85),
                  legend_background=element_rect(fill='none')))
p2.show()
#p2.save(plotnine(b).pdf)
```

```python
#-------------------------------(c) Position + Direction+Shape----------------------------------
p3=(ggplot(df,aes(x='Time',y='value',shape='variable'))+
       geom_line()+
       geom_point(size=4,color='k',fill='#BFBFBF')+
       scale_shape_manual(values=('o','s','D','^'))+
       theme_classic()+
       theme(text=element_text(size=12,color='k'),aspect_ratio=0.8,
                  dpi=100, figure_size=(4,4),
                  legend_position=(0.12,0.85),
                  legend_background=element_rect(fill='none')))
p3.show()
#p3.save(plotnine(c).pdf)
```

```python
from plotnine import *
from plotnine.data import mpg
t=(ggplot(mpg,aes('cty','hwy',fill='fl'))+
     geom_point(size=3, stroke=0.3, alpha=0.8, show_legend=False)+
     scale_fill_hue(s=0.9, l=0.65, h=0.0417,color_space='husl'))
t.show() 
```

```python
 #---------------------------Arrange based on variables in rows-------------------------------------------
t2 = (t + facet_grid('year ~ .')
        + theme_matplotlib()
        + theme(text=element_text(size=12, colour = 'black'),
                      strip_text=element_text(size=15, colour='black'),
                      strip_background=element_rect(color='k'),
                      aspect_ratio = 0.5,
                      dpi=100,
                      figure_size=(4,4)))
t2.show()
```

```python
 #---------------------------Arrange based on two variables in a matrix layout.-------------------------------
t3 = (t + facet_grid('year ~fl')
        + theme_matplotlib()
        + theme(text=element_text(size=12,color='black'),
                      strip_text=element_text(size=15,color='black'),
                      strip_background=element_rect(color='k'),
                      aspect_ratio =1,
                      dpi=100,
                      figure_size=(9,9)))
t3.show()
```

```python
#-----------------------------Arrange based on variables in a rectangular layout.-------------------------
t4 = (t + facet_wrap('~ fl')
        + theme_matplotlib()
        + theme(text=element_text(size=14, color='black'),
                       strip_text=element_text(size=20,color='black'),
                       strip_background=element_rect(color='k'),
                       aspect_ratio=1,
                       dpi=100,
                       figure_size=(9,9)))
t4.show()
```

```python
#------------------------------- Adjust the range of values on the X-axis and Y-axis. --------------------
t5 = (t + facet_grid('drv ~ fl',scales = 'free')
        + theme_matplotlib()
        + theme(text=element_text(size=12,color='black'),
                      strip_text=element_text(size=15,color='black'),
                      strip_background=element_rect(color='k'),
                      aspect_ratio=1,
                      dpi=100,
                      figure_size=(9,9)))
t5.show()
#t5.save('分面5.pdf')
```

```python
import pandas as pd
from plotnine import *

mydata = pd.DataFrame({'Cut': ["Fair", "Good", "Very Good", "Premium", "Ideal"], 'Price': [4300, 3800, 3950, 4700, 3500]})

(ggplot(mydata , aes(x='Cut', y='Price'))+
             geom_bar(stat="identity", width=0.8)+
              theme(figure_size=(3.2,3),dpi=300))
```

```python
import pandas as pd
from plotnine import *
from pandas.api.types import CategoricalDtype 

mydata = pd.DataFrame({'Cut': ["Fair", "Good", "Very Good", "Premium", "Ideal"], 'Price': [4300, 3800, 3950, 4700, 3500]})

Sort_data = mydata.sort_values(by='Price', ascending=False)

Sort_data['Cut'] = Sort_data['Cut'].astype(CategoricalDtype(categories =Sort_data['Cut'].values,ordered=True))

(ggplot(Sort_data, aes(x='Cut', y='Price'))+
             geom_bar(stat="identity", width=0.8)+
              theme(figure_size=(3.2,3),dpi=300))
```
